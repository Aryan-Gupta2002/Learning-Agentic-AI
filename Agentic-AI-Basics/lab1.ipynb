```python
from openai import OpenAI
import os
from openai import OpenAI

llm = OpenAI(
        api_key=os.getenv("NVIDIA_API_KEY"),
        base_url="https://integrate.api.nvidia.com/v1",
    )
```

```python
messages = [{"role": "user", "content": "Tell me a short fun fact"}]
messages
```

```python
response = llm.chat.completions.create(model="meta/llama-3.1-8b-instruct", messages=messages)
print(response.choices[0].message.content)
```

```python
question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
response = llm.chat.completions.create(model="meta/llama-3.1-8b-instruct", messages=messages)
responded_question = response.choices[0].message.content
print(responded_question)
```

### The below code cell is LLM Chaining as we used response(Output) from above as input to below

```python
messages = [{"role": "user", "content": responded_question}]
messages
```

```python
response = llm.chat.completions.create(model="meta/llama-3.1-8b-instruct", messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
message = f"""
Here is a question:
{responded_question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

print(message)
```

```python
messages = [{"role": "user", "content": message}]
response = llm.chat.completions.create(model="mistralai/mixtral-8x7b-instruct-v0.1", messages=messages)
print(response.choices[0].message.content)
```

```python
from IPython.display import Markdown, display

display(Markdown(response.choices[0].message.content))
```
